```python
# ReverseAddPalindrome] 
#Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số 
# đối xứng (palindrome). In ra số bước và giá trị palindrome thu được.

#VD số đối xứng : 121 1331 11311

def Reverse_Number(n):
    rev = 0
    while n > 0:
        n % 10
        rev = rev * 10 + n % 10
        n //= 10
    return rev

def special_Palindrome_Number(n):
    return n == Reverse_Number(n)

def Palindrome_Number(n):
    step = 0
    while not special_Palindrome_Number(n):
        step += 1
        n = n + Reverse_Number(n)
    return step, n
n = int(input())
print(Palindrome_Number(n))

        
```

```python
# ReverseAddPalindrome] 
#Nhập số nguyên dương n. Lặp lại phép toán n = n + reverse(n) cho đến khi n là số 
# đối xứng (palindrome). In ra số bước và giá trị palindrome thu được.

#VD số đối xứng : 121 1331 11311

def reverse(n):
    rev = 0
    while n > 0:
        rev = rev * 10 + n % 10
        n //= 10
    return rev
def special_palindrome(n):
    return n == reverse(n)
def palindrome(n):
    step = 0
    while not special_palindrome(n):
        n = n + reverse(n)
        step += 1
    return step, n

n = int(input())
print(palindrome(n))
        
    
```
