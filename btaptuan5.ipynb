```python
#W5A3
# Viết hàm kiểm tra một số có phải là số nguyên tố hay không?
import math
def Check_Prime_Number(n):
    if n == 1:
        return False
        return
    if n == 2:
        return True
        return
    else:
        for i in range(2, int(math.sqrt(n) + 1)):
            if n % i == 0:
                print(n, "ko phai la so nguyen to")
                return False
        return True

n = int(input())
for x in range(2, int(math.sqrt(n) + 1)):
    if x in Check_Prime_Number:
        print(x)
Check_Prime_Number(n)
    
```

```python
#W5A4 
#VIết hàm kiểm tra một số có phải số hoàn hảo hay không, xử lý trên dữ liệu kiểu số nguyên
# Số hoàn hảo là số mà ước của nó (ko tính nó) cộng vào bằng chính nó

import math
def Perfect_Number():
    n = int(input("nhap n:"))
    tong = 0
    for i in range(1, n):
        if n % i == 0:
            tong += i
    print(tong)
    if tong == n:
        print(n, "la so hoan hao")
    else:
        print(n, " ko phai la so hoan hao")
Perfect_Number()
```

```python
#Viết hàm tính giai thừa. Nhập vào số nguyên n, trả về giai thừa của n (n!)
#W5A6
def giaithua (n):
    ketqua = 1
    for i in range(1, n + 1):
        ketqua *= i
    return ketqua
giaithua(6)
```

```python
#W5A9 Viết hàm trả về tổng các chữ số của số nguyên dương n
def tongchuso(n):
    tong = 0
    while n != 0:
        i = n % 10
        n //= 10
        tong += i
    return tong
tongchuso(57895)
    
        
        
```
