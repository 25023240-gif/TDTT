```python
class Cylinder:
    def __init__(self, r, h):
        self.r = r
        self.h = h
    def surface(self):
        return 2 * 3.14 * self.r * (self.r + self.h)
    def volume(self):
        return 3.14 * self.r * self.r * self.h

r, h = map(float, input().split())
c = Cylinder(r, h)
print(f"{c.surface():.2f} {c.volume():.2f}")
```

```python
class Date:
    def __init__(self, d, m, y):
        self.d = d
        self.m = m
        self.y = y
    def leap(self):
        return self.y % 400 == 0 or (self.y % 4 == 0 and self.y % 100 != 0)
    def valid(self):
        md = [31, 29 if self.leap() else 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
        return 1 <= self.m <= 12 and 1 <= self.d <= md[self.m-1]
    def next_day(self):
        md = [31, 29 if self.leap() else 28, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
        self.d += 1
        if self.d > md[self.m-1]:
            self.d = 1
            self.m += 1
            if self.m > 12:
                self.m = 1
                self.y += 1

d, m, y = map(int, input().split('/'))
dt = Date(d, m, y)
if not dt.valid():
    print("INVALID")
else:
    dt.next_day()
    print(f"{dt.d:02d}/{dt.m:02d}/{dt.y}")
```

```python
import math

class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def is_origin(self):
        return self.x == 0 and self.y == 0
    def on_y(self):
        return self.x == 0 and self.y != 0
    def on_x(self):
        return self.y == 0 and self.x != 0
    def dist(self):
        return math.sqrt(self.x*self.x + self.y*self.y)

x, y = map(float, input().split())
p = Point(x, y)
print("YES" if p.is_origin() else "NO")
print("YES" if p.on_y() else "NO")
print("YES" if p.on_x() else "NO")
print(f"{p.dist():.2f}")
```

```python
class Calculator:
    def __init__(self, a, b):
        self.a = a
        self.b = b
    def add(self):
        return self.a + self.b
    def subtract(self):
        return self.a - self.b
    def product(self):
        return self.a * self.b
    def divide(self):
        return self.a / self.b
    def power(self):
        return self.a ** self.b
    def mod(self):
        return self.a % self.b
    def set_numbers(self, a, b):
        self.a = a
        self.b = b

a, b = map(float, input().split())
cal = Calculator(a, b)
while True:
    op = input()
    if op.lower() == "yes":
        break
    if op == "add":
        print(cal.add())
    elif op == "subtract":
        print(cal.subtract())
    elif op == "product":
        print(cal.product())
    elif op == "divide":
        print(cal.divide())
    elif op == "power":
        print(cal.power())
    elif op == "mod":
        print(cal.mod())
    elif op == "set":
        a, b = map(float, input().split())
        cal.set_numbers(a, b)
```

```python
class ShoppingCart:
    def __init__(self):
        self.items = []
    def add(self, name, price):
        self.items.append({"name": name, "price": price})
    def remove(self, name):
        self.items = [i for i in self.items if i["name"] != name]
    def empty(self):
        return len(self.items) == 0
    def total(self):
        return sum(i["price"] for i in self.items)
    def show(self):
        for i in self.items:
            print(i["name"], i["price"])
    def clear(self):
        self.items = []

cart = ShoppingCart()
while True:
    cmd = input()
    if cmd == "yes":
        break
    p = cmd.split()
    if p[0] == "add":
        cart.add(p[1], float(p[2]))
    elif p[0] == "remove":
        cart.remove(p[1])
    elif p[0] == "total":
        print(cart.total())
    elif p[0] == "show":
        cart.show()
    elif p[0] == "clear":
        cart.clear()
```

```python
class Rectangle:
    def __init__(self, w, h):
        self.w = w
        self.h = h
    def area(self):
        return self.w * self.h
    def perimeter(self):
        return 2 * (self.w + self.h)
    def scale(self, k):
        self.w *= k
        self.h *= k

w, h, k = map(float, input().split())
r = Rectangle(w, h)
r.scale(k)
print(f"{r.area():.2f} {r.perimeter():.2f}")
```

```python
import math

class Fraction:
    def __init__(self, n, d):
        if d < 0:
            n, d = -n, -d
        g = math.gcd(n, d)
        self.n = n // g
        self.d = d // g
    def add(self, o):
        return Fraction(self.n*o.d + o.n*self.d, self.d*o.d)
    def sub(self, o):
        return Fraction(self.n*o.d - o.n*self.d, self.d*o.d)
    def mul(self, o):
        return Fraction(self.n*o.n, self.d*o.d)
    def div(self, o):
        return Fraction(self.n*o.d, self.d*o.n)
    def __str__(self):
        return f"{self.n}/{self.d}"

a, b, op, c, d = input().split()
f1 = Fraction(int(a), int(b))
f2 = Fraction(int(c), int(d))
if op == "+":
    print(f1.add(f2))
elif op == "-":
    print(f1.sub(f2))
elif op == "*":
    print(f1.mul(f2))
else:
    print(f1.div(f2))
```

```python
class Student:
    def __init__(self, name):
        self.name = name
        self.scores = []
    def add_score(self, s):
        self.scores.append(s)
    def avg(self):
        return sum(self.scores) / len(self.scores)
    def rank(self):
        a = self.avg()
        if a >= 8:
            return "Excellent"
        if a >= 6.5:
            return "Good"
        if a >= 5:
            return "Average"
        return "Poor"

name = input()
st = Student(name)
n = int(input())
for _ in range(n):
    _, s = input().split()
    st.add_score(float(s))
print(f"{st.name} {st.avg():.2f} {st.rank()}")
```

```python
class Book:
    def __init__(self, t, a, y):
        self.t = t
        self.a = a
        self.y = y

class Library:
    def __init__(self):
        self.books = []
    def add(self, b):
        self.books.append(b)
    def count_by_author(self, a):
        return sum(1 for b in self.books if b.a == a)
    def find_by_year(self, y):
        return sum(1 for b in self.books if b.y == y)

lib = Library()
n = int(input())
for _ in range(n):
    s = input()
    if s.startswith("ADD"):
        _, x = s.split(" ", 1)
        t, a, y = x.split(";")
        lib.add(Book(t, a, int(y)))
    elif s.startswith("COUNTYEAR"):
        print(lib.find_by_year(int(s.split()[1])))
    else:
        print(lib.count_by_author(s.split()[1]))
```

```python
import math

class Point3D:
    def __init__(self, x, y, z):
        self.x = x
        self.y = y
        self.z = z
    def vec(self, o):
        return (o.x-self.x, o.y-self.y, o.z-self.z)

def cross(u, v):
    return (u[1]*v[2]-u[2]*v[1], u[2]*v[0]-u[0]*v[2], u[0]*v[1]-u[1]*v[0])

def dot(u, v):
    return u[0]*v[0] + u[1]*v[1] + u[2]*v[2]

A = Point3D(*map(float, input().split()))
B = Point3D(*map(float, input().split()))
C = Point3D(*map(float, input().split()))
D = Point3D(*map(float, input().split()))

n1 = cross(A.vec(B), A.vec(C))
n2 = cross(B.vec(C), B.vec(D))

cos = dot(n1, n2) / (math.sqrt(dot(n1, n1)) * math.sqrt(dot(n2, n2)))
phi = math.degrees(math.acos(cos))
print(f"{phi:.2f}")
```

```python
class ComplexNumber:
    def __init__(self, r, i):
        self.r = r
        self.i = i
    def multiply(self, o):
        return ComplexNumber(self.r*o.r - self.i*o.i, self.r*o.i + self.i*o.r)

a, b = map(float, input().split())
c, d = map(float, input().split())
z1 = ComplexNumber(a, b)
z2 = ComplexNumber(c, d)
z = z1.multiply(z2)
print(z.r, z.i)
```

```python
import math

class Triangle:
    def __init__(self, a, b, c):
        self.a = a
        self.b = b
        self.c = c
    def area(self):
        if self.a + self.b <= self.c or self.a + self.c <= self.b or self.b + self.c <= self.a:
            return -1
        p = (self.a + self.b + self.c) / 2
        return math.sqrt(p*(p-self.a)*(p-self.b)*(p-self.c))

a, b, c = map(int, input().split())
t = Triangle(a, b, c)
res = t.area()
if res < 0:
    print("invalid")
else:
    print(f"{res:.2f}")
```
