```python
# Cho một mảng số nguyên không âm. Gọi
# mảng đó là "mảng đẹp" nếu mỗi phần tử
# x trong mảng xuất hiện đúng x lần.
# Kiểm tra xem mảng có phải là mảng đẹp
# hay không. Nếu không hãy xóa đi y số (y
# là nhỏ nhất có thể) để mảng còn lại thành
#mảng đẹp

def MangDep(arr):
    count = {}
    for i in arr:
        if i not in count:
            count[i] = 1
        else:
            count[i] += 1
    y = 0
    for x in count:
        if count[x] == x:
            continue
        elif count[x] < x:
            y += count[x]
        elif count[x] > x:
            y += count[x] - x
    if y == 0:
        return "YES"
    else:
        return f"NO {y}"
arr = list(map(int,input().split()))
MangDep(arr)     
```

```python
# Tìm phần tử xuất hiện nhiều hơn n/2 lần
# trong mảng (nếu có) với độ phức tạp thời
# gian là O(n), độ phức tạp bộ nhớ là O(1),
# - Boyer–Moore. Nếu không có phần tử
# nào thỏa mãn thì không in gì

arr = list(map(int,input().split()))
n = len(arr)
dic = {}
for i in arr:
    if i not in dic:
        dic[i] = 1
    else:
        dic[i] += 1
for x in dic:
    if dic[x] > n/2:
        print(x)
```

```python
# Sắp xếp có điều kiện: Số chẵn đứng
# trước, số lẻ đứng sau, chẵn tăng dần, lẻ
# giảm dần

arr = list(map(int,input().split()))
a = []
b = []
for i in arr:
    if i % 2 == 0:
        a.append(i)
    else:
        b.append(i)
a.sort()
b.sort(reverse=True)
result = a + b
print(*result,sep = " ")
```

```python
# Merge 2 mảng đã sắp xếp (có thể tăng
# dần, có thể giảm dần) mà không dùng
# bất cứ thuật toán sắp xếp nào thành một
# mảng cũng có cùng thứ tự sắp xếp như 2
# mảng ban đầu

lst_1 = list(map(int,input().split()))
lst_2 = list(map(int,input().split()))
a = lst_1 + lst_2
a.sort()
print(*a, sep = " ")
```

```python
# Cho dãy N số nguyên dương, tính tổng   
# các số có k chữ số. Dòng đầu ghi N số
# nguyên dương, dòng sau ghi số nguyên
# dương k

def DemChuSo(n):
    count = 0
    while n > 0:
        count += 1
        n //= 10
    return count

lst = list(map(int,input().split()))
k = int(input())
tong = 0 
for i in lst:
    if DemChuSo(i) == k:
        tong += i
print(tong)
```

```python
# Cho N là ngày, T là tháng năm 2020,
# Tính xem đó là ngày thứ bao nhiêu của
# năm đó nếu đó là ngảy hợp lệ
Days_of_2020 = [31, 29, 31, 30, 31, 30, 31, 31, 30, 31, 30, 31]
def ngay_thu_bao_nhieu(N, T):
    if T > 12 or T < 0:
        return "Invalid"
    elif N < 0 or N > Days_of_2020[T - 1]:
        return "Invalid"
    else:
        Y = sum(Days_of_2020[:T - 1]) + N
        return Y

N, T = map(int,input().split())
ngay_thu_bao_nhieu(N, T)
    
    
```

```python
#### Cho N tên sinh viên gồm cả họ và tên,
# tìm xem tên nào hay gặp nhất. Nếu có
# nhiều tên cùng số lần gặp thì lấy tên có
# thứ tự từ điển nhỏ nhất

N = int(input())
name = []
for _ in range (N):
    full_name = input().split()
    name.append(full_name[-1])
dic = {}
for i in name:
    if i not in dic:
        dic[i] = 1
    else:
        dic[i] += 1
max_count = 0
result = ""
for x in dic:
    if dic[x] > max_count:
        max_count = dic[x]
        result = x
print(result)
```

```python
# Đếm số lượng tên trong xâu kí tự. Tên
# được định nghĩa là xâu có độ dài >=2 và
# có chữ cái đầu viết hoa, các chữ cái khác
# viết thường, chỉ có kí tự cái.

s = input().split()
result = []
for x in s:
    if x not in result:
        result.append(x)
count = 0
for i in result:
    if len(i) >= 2 and i[0].isupper():
        count += 1
print(count)
```
