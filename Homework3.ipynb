```python
#bai9
#Bài 9: Nhập vào số nguyên dương a, kiểm tra xem a có phải là số chính phương hay
#không? Số chính phương là số có căn bậc hai là một số nguyên VD: 1, 4, 9, 16, 25…. 

a = int(input("nhap vao so nguyen duong"))
square_a =  False
for i in range(1, a + 1):
    if i*i == a:
        square_a = True
        break
if square_a == True:
    print(a, "khong phai la so chinh phuong")
else:
    print(a, "la so chinh phuong")
    
```

```python
#bai10
#Bài 10: Nhập vào số nguyên dương a, kiểm tra xem a có phải là số hoàn hảo hay không.
#Số hoàn hảo là số có tổng các ước (ước nhỏ hơn nó) bằng chính nó, VD: số 6, 28… 

a = int(input("nhap vao so nguyen duong"))
S = 0
for i in range(a,a + 1):
    if a % i == 0:
        S += i
if S == a:
    print(a, "la so hoan hao")
else:
    print(a, "khong phai la so hoan hao")

```

```python
#bai12
#Nhập số nguyên dương n, Cho S(k) = 1 + 2 + 3 + … + k. Tìm k sao cho S(k)
#lớn nhất nhưng nhỏ hơn n

n = int(input("nhap vao so nguyen duong n:"))
for k in range (1, n + 1):
    S = (k * (k + 1))/2
    if S > n:
        print(k - 1)
        break
```

```python
#Bài 15: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số. Làm bằng nhiều 
#        cách nhất có thể, ko dùng mảng.

# cach 1
n = int(input("nhap vao so nguyen duong n"))
count = 0
for _ in range (50):
    if n == 0:
        break
    n //= 10
    count += 1 
print("so chu so la",count)
    

```

```python
#Bài 16: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số chẵn, bao nhiêu 
#        chữ số lẻ


n = int(input("nhap vao so nguyen duong n:"))
chan = 0
le = 0 
for _ in range(50):
    if n == 0:
        break
    n % 10  #lay chu so cuoi
    if n % 2 == 0:
        chan += 1
    if n % 2 == 1:
        le += 1 
    n //= 10  #bo chu so cuoi
print(" n co so chu so chan la: ", chan)
print(" n co so chu so le la: " , le)
 
        
```

```python
#bai 17 Nhập vào số nguyên dương n, tính tổng các chữ số của n


n = int(input(" nhap vao so nguyen duong n:"))
s = 0
for _ in range (50):
    if n == 0:
        break
    t = n % 10 
    s = s + t
    n //= 10 
print ( " n co tong cac chu so la: ", s)
    
```

```python
#Bài 20: Nhập vào số nguyên dương n, đếm xem n có bao nhiêu chữ số là số nguyên tố.

n = int(input("nhap vao so nguyen duong n"))
count = 0
s = n
for _ in range(100):
    if s == 0:
        break
    digit = s % 10 
    if digit in (2, 3 ,5 ,7):
        count += 1
    s //= 10
print(" n co so chu so nguyen to la:", count)
    
    
```
