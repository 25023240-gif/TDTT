```python
a = list(map(int, input().split()))
b = list(map(int, input().split()))
i = j = 0
res = []
while i < len(a) and j < len(b):
    if a[i] <= b[j]:
        res.append(a[i])
        i += 1
    else:
        res.append(b[j])
        j += 1
res += a[i:]
res += b[j:]
print(*res)
```

```python
import heapq

mat = []
while True:
    try:
        row = list(map(int, input().split()))
        if len(row) == 1:
            k = row[0]
            break
        mat.append(row)
    except:
        break

n = len(mat)
m = len(mat[0])
h = []
for i in range(n):
    heapq.heappush(h, (mat[i][0], i, 0))

cnt = 0
while True:
    val, r, c = heapq.heappop(h)
    cnt += 1
    if cnt == k:
        print(val)
        break
    if c + 1 < m:
        heapq.heappush(h, (mat[r][c+1], r, c+1))
```

```python
s = list(input().strip())
n = len(s)
best = ''.join(s)
for i in range(n):
    for j in range(i+1, n):
        if i == 0 and s[j] == '0':
            continue
        s[i], s[j] = s[j], s[i]
        best = min(best, ''.join(s))
        s[i], s[j] = s[j], s[i]
print(best)
```

```python
def can(a, b):
    i = j = 0
    while i < len(a):
        if j < len(b) and a[i].upper() == b[j]:
            j += 1
        elif a[i].islower():
            pass
        else:
            return False
        i += 1
    return j == len(b)

q = int(input())
for _ in range(q):
    a = input().strip()
    b = input().strip()
    print("YES" if can(a, b) else "NO")
```

```python
n = int(input())
p = list(map(int, input().split()))

i = n - 2
while i >= 0 and p[i] >= p[i+1]:
    i -= 1

if i < 0:
    print(*sorted(p))
else:
    j = n - 1
    while p[j] <= p[i]:
        j -= 1
    p[i], p[j] = p[j], p[i]
    p[i+1:] = reversed(p[i+1:])
    print(*p)
```

```python
bed = list(map(int, input().split()))
k = int(input())

i = 0
while i < len(bed):
    if bed[i] == 0:
        left = (i == 0 or bed[i-1] == 0)
        right = (i == len(bed)-1 or bed[i+1] == 0)
        if left and right:
            bed[i] = 1
            k -= 1
            if k == 0:
                break
            i += 1
    i += 1

print(k <= 0)
```

```python
x = y = 0
seen = {(0, 0)}
moves = input().strip()

for c in moves:
    if c == 'R':
        x += 1
    elif c == 'L':
        x -= 1
    elif c == 'U':
        y += 1
    else:
        y -= 1
    if (x, y) in seen:
        print(True)
        break
    seen.add((x, y))
else:
    print(False)
```

```python
def Median(a):
    x1, x2, x3, x4, x5 = a
    if x1 > x2: x1, x2 = x2, x1
    if x3 > x4: x3, x4 = x4, x3
    if x1 > x3: x1, x3 = x3, x1
    if x2 > x4: x2, x4 = x4, x2
    if x2 > x3: x2, x3 = x3, x2
    return x3
```

```python
import bisect

houses = list(map(int, input().split()))
heaters = list(map(int, input().split()))
houses.sort()
heaters.sort()

ans = 0
for h in houses:
    i = bisect.bisect_left(heaters, h)
    d1 = abs(h - heaters[i]) if i < len(heaters) else 10**18
    d2 = abs(h - heaters[i-1]) if i > 0 else 10**18
    ans = max(ans, min(d1, d2))

print(ans)
```
