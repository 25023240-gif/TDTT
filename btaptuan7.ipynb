```python
#W7A1
# Viết hàm tìm kiếm nhị phân
# trên một danh sách đã
# được sắp xếp tăng dần.
# Trả về chỉ số của phần tử
# hoặc -1 nếu không tìm
# thấy. mẫu hàm: def
# binary_search(arr, left,
# right, target):

def Binary_search(arr, value):
    left, right = 0, len(arr) - 1
    while left <= right:
        mid = (left + right)//2
        if arr[mid] == value:
            return mid
        elif arr[mid] < value:
            left = mid + 1
        else : 
            right = mid - 1
    return -1

lst = list(map(int,input().split()))
k = int(input())
Binary_search(lst,k)

```

```python
#W7A2
# Cho danh sách arr và một
# số nguyên x, đếm xem x
# xuất hiện bao nhiêu lần
# trong danh sách. def
# count_occurrences(arr, x) -
# > int:

def count_occurrences(arr,x):
    count = 0
    for i in arr:
        if i == x:
            count += 1
    return count
lst = list(map(int,input().split()))
k = int(input())
count_occurrences(lst,k)
            
        
        
```

```python
#W7A3
# Viết hàm thực hiện thuật toán bubble sort, Selection Sort để sắp xếp danh sách theo thứ tự tăng dần.
def bubble_sort(arr):
    n = len(arr)
    for i in range(n):
        swapped = False
        for j in range(0, n - i - 1):
            if arr[j] > arr[j + 1]:
                arr[j], arr[j + 1] = arr[j + 1], arr[j]
                swapped = True
        if not swapped:
            break
lst = list(map(int,input().split()))
bubble_sort(lst)
print(lst)
                

                
                
            
```

```python
#W7A3
def selection_sort(arr):
    n = len(arr)
    for i in range(n):
        min_idx = i
        for j in range(i + 1 , n):
            if arr[j] < arr[min_idx]:
                min_idx = j
            arr[i], arr[min_idx] = arr[min_idx], arr[i]
lst = list(map(int,input().split()))
selection_sort(lst)
print(lst)
    
    
    
```

```python
#W7A4 
# Viết hàm tìm phần tử có số lần xuất hiện nhiều nhất trong danh sách. Nếu có nhiều phần tử cùng số lần, trả về phần tử xuất hiện trước.
def tim_ptu_nhieunhat_somnhat(arr):
    dic = {}
    for x in arr:
        if x not in dic:
            dic[x] = 1
        else:
            dic[x] += 1
    nhieu_nhat = 0
    ketqua = 0
    if x > nhieu_nhat:
        nhieu_nhat = dic[x]
        ketqua = x
    return ketqua, nhieu_nhat
lst = list(map(int,input().split()))
value, times = tim_ptu_nhieunhat_somnhat(lst)
print(f"{value} xuất hiện nhiều nhất, sớm nhất, {times} lần")

            
        
```

```python
#W7A5
# Cho danh sách các số
# nguyên và một số X, đếm
# tất cả các cặp (a, b) sao
# cho a + b = X

lst = list(map(int,input().split()))
n = len(lst)
x = int(input())
count = 0
for i in range(n):
    for j in range(i + 1, n):
        if lst[i] + lst[j] == x:
            count += 1  
print(count)
    
```
