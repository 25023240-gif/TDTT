```python
#W11A4
def areAnagrams(path1: str, path2: str) -> bool:
    with open(path1, "r", encoding="utf-8") as f1:
        s1 = f1.read().lower()
    with open(path2, "r", encoding="utf-8") as f2:
        s2 = f2.read().lower()

    s1 = ''.join(c for c in s1 if c.isalpha())
    s2 = ''.join(c for c in s2 if c.isalpha())

    return sorted(s1) == sorted(s2)


with open("a.txt", "w", encoding="utf-8") as f:
    f.write("Tom Marvolo Riddle")

with open("b.txt", "w", encoding="utf-8") as f:
    f.write("I Am Lord Voldemort.")

print(areAnagrams("a.txt", "b.txt"))
```

```python
#W11A5
def count(path: str, character: str) -> int:
    with open(path, "r", encoding="utf-8") as f:
        sentence = f.read()
    return sentence.count(character)


with open("input.txt", "w", encoding="utf-8") as f:
    f.write("hello world")

print(count("input.txt", "l"))
```

```python
#W11A6
def countCharacter(path: str, letter: str) -> int:
    with open(path, "r", encoding="utf-8") as f:
        text = f.read().lower()
    return text.count(letter.lower())


with open("text.txt", "w", encoding="utf-8") as f:
    f.write("Everyone deserves a second shot.")

print(countCharacter("text.txt", "e"))
```

```python
#W11A10
def getMoney(path: str) -> list[int]:
    with open(path, "r") as f:
        return list(map(int, f.read().strip().split()))


def rob(hourses: list[int]) -> int:
    if not hourses:
        return 0
    if len(hourses) == 1:
        return hourses[0]

    dp0 = hourses[0]
    dp1 = max(hourses[0], hourses[1])

    for i in range(2, len(hourses)):
        dp0, dp1 = dp1, max(dp1, dp0 + hourses[i])

    return dp1


with open("money.txt", "w") as f:
    f.write("2 7 9 3 1")

houses = getMoney("money.txt")
print(rob(houses))
```

```python
#W11A2
def maximumProduct(path: str) -> int:
    with open(path, "r") as f:
        n = int(f.readline().strip())
        nums = list(map(int, f.readline().split()))
    nums.sort()
    return max(nums[-1] * nums[-2] * nums[-3],
               nums[0] * nums[1] * nums[-1])


with open("input.txt", "w") as f:
    f.write("5\n-10 -10 1 3 2")

print(maximumProduct("input.txt"))
```

```python
#W11A13
def moveZeroes(fileName: str) -> list:
    with open(fileName, "r") as f:
        n = int(f.readline().strip())
        nums = list(map(int, f.readline().split()))

    res = []
    c = 0
    for x in nums:
        if x == 0:
            c += 1
        else:
            res.append(x)
    res.extend([0] * c)
    return res


with open("input.txt", "w") as f:
    f.write("7\n0 1 0 3 12 0 5")

print(moveZeroes("input.txt"))
```

```python
#W11A14
def averageTime(path: str):
    with open(path, "r") as f:
        f.readline()
        total = 0
        cnt = 0
        for line in f:
            total += float(line.strip().split(",")[1])
            cnt += 1
    return round(total / cnt, 2)


with open("laps.csv", "w") as f:
    f.write(
        "Lap,Time\n"
        "1,143.23\n"
        "2,134.12\n"
        "3,150.50\n"
    )

print(averageTime("laps.csv"))
```

```python
#W11A15
def findMovies(path: str) -> list[str]:
    res = []
    with open(path, "r", encoding="utf-8") as f:
        f.readline()
        for line in f:
            parts = [p.strip() for p in line.split(",")]
            if float(parts[3]) > 8.0 and parts[2].lower() != "boring":
                res.append(parts[1])
    if not res:
        return ["All are boring!"]
    return res


with open("movies.csv", "w", encoding="utf-8") as f:
    f.write(
        "id, movie, description, rating\n"
        "1, Thor, great 3D, 8.0\n"
        "2, The Rings, Fantasy, 7.5\n"
        "3, Irish, boring, 6.2\n"
        "4, Ice song, Fantasy, 8.6\n"
        "5, Home Alone, Interesting, 6.5\n"
    )

print(findMovies("movies.csv"))
```

```python
#W11A16
import math

with open("shapes.txt", "w") as f:
    f.write(
        "TYPE: SQUARE\n"
        "SIDE: 5.3\n\n"
        "TYPE: RECTANGLE\n"
        "WIDTH: 5.0 HEIGHT: 16.0\n\n"
        "TYPE: CIRCLE\n"
        "RADIUS: 2.15\n"
    )

path = "shapes.txt"

with open(path, "r") as f:
    lines = [line.strip() for line in f if line.strip() != ""]

i = 0
while i < len(lines):
    shape = lines[i].split(":")[1].strip()
    data = lines[i + 1]

    if shape == "SQUARE":
        side = float(data.split(":")[1])
        p = 4 * side
    elif shape == "RECTANGLE":
        parts = data.replace(":", "").split()
        w = float(parts[1])
        h = float(parts[3])
        p = 2 * (w + h)
    else:
        r = float(data.split(":")[1])
        p = 2 * math.pi * r

    print(f"{p:.2f}")
    i += 2
```

```python
#W11A17
def productExceptSelf(path: str) -> list[int]:
    with open(path, "r") as f:
        nums = list(map(int, f.read().split()))

    n = len(nums)
    res = [1] * n

    left = 1
    for i in range(n):
        res[i] = left
        left *= nums[i]

    right = 1
    for i in range(n - 1, -1, -1):
        res[i] *= right
        right *= nums[i]

    return res


with open("input.txt", "w") as f:
    f.write("1 2 3 4")

print(productExceptSelf("input.txt"))
```

```python
#W11A18
import math

def estimatedTime(path: str):
    with open(path, "r") as f:
        src = f.readline().replace(":", "").split()
        des = f.readline().replace(":", "").split()
        vel = f.readline().replace(":", "").split()

    x1, y1 = float(src[2]), float(src[4])
    x2, y2 = float(des[2]), float(des[4])
    v = float(vel[1])

    dist = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)
    return round(dist / v, 2)


with open("missile.txt", "w") as f:
    f.write(
        "SRC X: 0 Y: 0\n"
        "DES X: 10000 Y: 10000\n"
        "VELOCITY: 6 KM/S\n"
    )

print(estimatedTime("missile.txt"))
```

```python
import struct

with open("data.bin", "wb") as f:
    f.write(struct.pack("ddd", 3.1415926535, 2.7182818284, 1.4142135623))

path = input().strip()

with open(path, "rb") as f:
    while True:
        bytes_data = f.read(8)
        if not bytes_data:
            break
        x = struct.unpack("d", bytes_data)[0]
        print(f"{x:.5f}")
```

```python
#W11A20
def imageSmoother(path: str):
    with open(path, "r") as f:
        M = int(f.readline().strip())
        img = [list(map(int, f.readline().split())) for _ in range(M)]

    res = [[0]*M for _ in range(M)]
    dirs = [-1, 0, 1]

    for i in range(M):
        for j in range(M):
            s = 0
            cnt = 0
            for dx in dirs:
                for dy in dirs:
                    ni, nj = i + dx, j + dy
                    if 0 <= ni < M and 0 <= nj < M:
                        s += img[ni][nj]
                        cnt += 1
            res[i][j] = s // cnt

    for row in res:
        print(*row)


with open("image.txt", "w") as f:
    f.write(
        "3\n"
        "100 200 100\n"
        "200 50 200\n"
        "100 200 100\n"
    )

imageSmoother("image.txt")
```
